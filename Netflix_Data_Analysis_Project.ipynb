# Netflix Data Analysis Project

## Introduction
This project analyzes Netflix movies and TV shows dataset to understand content trends, genres, and distribution across countries using Python.

```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Load Dataset

```python
df = pd.read_csv("netflix_titles.csv")
df.head()
```

## Data Overview

 Movies vs TV Shows Analysis

```python
df['type'].value_counts()
```

```python
df['type'].value_counts().plot(kind='bar')
plt.title("Movies vs TV Shows on Netflix")
plt.show()
```

 Top Countries Producing Content

```python
df['country'].value_counts().head(10)
```

```python
df['country'].value_counts().head(10).plot(kind='bar')
plt.title("Top Countries on Netflix")
plt.show()
```

 Most Common Genres

```python
df['listed_in'].value_counts().head(10)
```

```python
df['listed_in'].value_counts().head(10).plot(kind='bar')
plt.title("Top Genres on Netflix")
plt.show()
```

# Content Trend Over Years

```python
df['release_year'].value_counts().sort_index().plot(kind='line')
plt.title("Netflix Content Over Years")
plt.show()
```

## Key Insights
- Netflix has more movies than TV shows  
- United States produces most content  
- Drama is the most common genre  
- Content increased after 2015
